## Step 2

Import the required libraries.

```python
import torch

from diffusers import StableDiffusionPipeline

from PIL import Image

import matplotlib.pyplot as plt

import os
```

## Step 3

Check whether GPU is available.

```python
device = "cuda" if torch.cuda.is_available() else "cpu"

print("Using Device:", device)
```

## Step 4

Load the pre-trained Stable Diffusion model.

```python
model_id = "runwayml/stable-diffusion-v1-5"

pipe = StableDiffusionPipeline.from_pretrained(
    model_id,
    torch_dtype=torch.float16
)

pipe = pipe.to(device)
```

## Step 5

Create a folder to save generated images.

```python
os.makedirs("outputs", exist_ok=True)
```

## Step 6

Generate an image from a text prompt.

```python
prompt = "An astronaut riding a horse on the moon"

image = pipe(prompt).images[0]

image
```

## Step 7

Display the generated image.

```python
plt.figure(figsize=(8,8))

plt.imshow(image)

plt.axis("off")

plt.show()
```

## Step 8

Save the generated image.

```python
image.save("outputs/astronaut_horse.png")

print("Image Saved Successfully")
```

## Step 9

Generate multiple images using different prompts.

```python
prompts = [
    "A futuristic city at sunset",
    "A fantasy castle on floating islands",
    "A robot painting a landscape",
    "A cute baby dragon reading books"
]
```

```python
for i, prompt in enumerate(prompts):

    image = pipe(prompt).images[0]

    image.save(f"outputs/image_{i+1}.png")

    plt.figure(figsize=(6,6))

    plt.imshow(image)

    plt.title(prompt)

    plt.axis("off")

    plt.show()
```

## Conclusion

Successfully generated images from text prompts using the Stable Diffusion pre-trained model.

The generated images are automatically saved in the **outputs** folder.
